Find the factorial of a number.

```python
x=int(input())
fact=1
print(f"The factorial of {x} is : ",end="")
if x==0 or x==1:
  pass
else:
  for i in range(x,0,-1):
    fact=fact*i
print(fact)
```

Count the number of digits in a number.

```python
num=int(input())
count=0
while num:
  count+=1
  num=num//10
print("Number of digits:", count)
```

Find the sum of digits of a number.

```python
nums=int(input())
sum_digits=0
while nums!=0:
  sum_digits+=nums%10
  nums=nums//10
print("Sum of digits:", sum_digits)
```

Check whether a number is a palindrome.

```python
n=int(input())
co=n
rev=""
while n!=0:
  new_num=n%10
  rev+=str(new_num)
  n=n//10

print( "Its a palindrome" if co==int(rev) else print("Its not a plindrome"))
```
