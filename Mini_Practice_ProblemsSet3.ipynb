Generate the Fibonacci series up to N term

```python
first=0
second=1
number=int(input())
print(first,end="->")
print(second,end="->")
for i in range(number):
  x=first+second
  print(x,end="->")
  first=second
  second=x
```

Check whether a number is an Armstrong number.


An Armstrong number is a number where the sum of each digit raised to the power of the number of digits equals the original number.


```python
num = int(input())
original = num
digits = len(str(num))
total = 0

while num != 0:
    digit = num % 10
    total += digit ** digits
    num //= 10

if total == original:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

Check whether a number is a perfect number.


A perfect number is a positive number that is equal to the sum of its proper divisors.

```python
num = int(input("Enter a number: "))

total = 0

for i in range(1, num):
    if num % i == 0:
        total += i

if total == num:
    print("Perfect number")
else:
    print("Not a perfect number")
```

Find the **second-largest digit** in a number.  

```python
num = int(input("Enter a number: "))

largest = -1
second_largest = -1

while num != 0:
    digit = num % 10

    if digit > largest:
        second_largest = largest
        largest = digit
    elif digit > second_largest and digit != largest:
        second_largest = digit

    num //= 10

if second_largest == -1:
    print("No second-largest digit")
else:
    print("Second-largest digit is", second_largest)
```
