```python
n = 5

for i in range(n):
    for j in range(i + 1):
        if j % 2 == 0:
            print(1, end=" ")
        else:
            print(0, end=" ")
    print()
```

```python


triangle = []

for i in range(n):
    row = [1] * (i + 1)

    for j in range(1, i):
        row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]

    triangle.append(row)

for row in triangle:
    print(" " * (n - len(row)), end="")

    for num in row:
        print(num, end=" ")

    print()
```
