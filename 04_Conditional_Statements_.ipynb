if Statement

The if statement executes code only when a condition is True.

```python
age = 20

if age >= 18:
    print("Eligible")
```

if-else

use else when there are two possible outcomes.

if condition

     
     "↓"

True → if block

False → else block

```python
age = 16

if age >= 18:
    print("Eligible")
else:
    print("Not eligible")
```

if-elif-else

Use elif when there are multiple conditions

Python checks conditions from top to bottom.

```python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("D")
```

```python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 60:
    print("C")
else:
    print("D")
```

```python
age = 20
has_id = True

if age >= 18 and has_id:
    print("Entry allowed")
```

Positive or negative

```python
number = int(input("Enter a number: "))

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```

Even or odd

```python
number = int(input("Enter a number: "))

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

Divisibility

```python
number = int(input("Enter a number: "))

if number % 5 == 0:
    print("Divisible by 5")
else:
    print("Not divisible by 5")
```

Character Conditions

```python
character = input("Enter a character: ")

if character == "a":
    print("You entered a")
else:
    print("Different character")
```

Vowel/Consonants

```python
character = input("Enter a character: ")

if character in "aeiou":
    print("Vowel")
else:
    print("Consonant")
```

Check whether a number falls within a range.

```python
number = int(input("Enter a number: "))

if number >= 1 and number <= 100:
    print("Number is between 1 and 100")
else:
    print("Number is outside the range")
```

Discount

```python
amount = float(input("Enter purchase amount: "))

if amount >= 5000:
    discount = amount * 0.20
elif amount >= 2000:
    discount = amount * 0.10
else:
    discount = 0

final_amount = amount - discount

print("Final amount:", final_amount)
```
