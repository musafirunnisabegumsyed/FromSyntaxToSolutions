.


```python
#Total bill when price and quantity are given
price = float(input("Enter the price: "))
quantity = int(input("Enter the quantity: "))

total_bill = price * quantity

print(f"Total Bill: ₹{total_bill:.2f}")
```

```python
#Bill for 3 different products
price1 = float(input("Enter price of product 1: "))
quantity1 = int(input("Enter quantity of product 1: "))

price2 = float(input("Enter price of product 2: "))
quantity2 = int(input("Enter quantity of product 2: "))

price3 = float(input("Enter price of product 3: "))
quantity3 = int(input("Enter quantity of product 3: "))

total1 = price1 * quantity1
total2 = price2 * quantity2
total3 = price3 * quantity3

total_bill = total1 + total2 + total3

print(f"Total Bill: ₹{total_bill:.2f}")
```

```python
#Total price after applying a discount
price = float(input("Enter the price: "))
discount = float(input("Enter discount percentage: "))

discount_amount = price * discount / 100
final_price = price - discount_amount

print(f"Discount Amount: ₹{discount_amount:.2f}")
print(f"Final Price: ₹{final_price:.2f}")
```

```python
#Calculate GST on a product price
price = float(input("Enter the product price: "))
gst_rate = float(input("Enter GST percentage: "))

gst = price * gst_rate / 100

print(f"GST Amount: ₹{gst:.2f}")
```

```python
#Calculate final bill including GST
price = float(input("Enter the product price: "))
gst_rate = float(input("Enter GST percentage: "))

gst = price * gst_rate / 100
final_bill = price + gst

print(f"GST Amount: ₹{gst:.2f}")
print(f"Final Bill: ₹{final_bill:.2f}")
```

Calculate the area and perimeter of a circle.

Calculate simple interest and total amount.

Convert Fahrenheit to Celsius.

Calculate a student's total marks, average, and percentage.

Calculate salary after adding a bonus.

Calculate monthly salary from annual salary.

Calculate total cost = price × quantity + tax
.
Calculate the distance travelled = speed × time.

Calculate the BMI using weight and height.


```python
# 1. Area and perimeter of a circle

import math

radius = float(input("Enter radius of the circle: "))

area = math.pi * radius ** 2
perimeter = 2 * math.pi * radius

print(f"Area of circle: {area:.2f}")
print(f"Perimeter of circle: {perimeter:.2f}")


# 2. Simple interest and total amount

principal = float(input("\nEnter principal amount: "))
rate = float(input("Enter interest rate: "))
time = float(input("Enter time in years: "))

simple_interest = (principal * rate * time) / 100
total_amount = principal + simple_interest

print(f"Simple Interest: {simple_interest:.2f}")
print(f"Total Amount: {total_amount:.2f}")


# 3. Celsius to Fahrenheit

celsius = float(input("\nEnter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

print(f"Temperature in Fahrenheit: {fahrenheit:.2f}")


# 4. Fahrenheit to Celsius

fahrenheit = float(input("\nEnter temperature in Fahrenheit: "))

celsius = (fahrenheit - 32) * 5 / 9

print(f"Temperature in Celsius: {celsius:.2f}")


# 5. Total marks, average and percentage

mark1 = float(input("\nEnter marks for subject 1: "))
mark2 = float(input("Enter marks for subject 2: "))
mark3 = float(input("Enter marks for subject 3: "))

total_marks = mark1 + mark2 + mark3
average = total_marks / 3
percentage = (total_marks / 300) * 100

print(f"Total Marks: {total_marks:.2f}")
print(f"Average: {average:.2f}")
print(f"Percentage: {percentage:.2f}%")


# 6. Salary after adding bonus

salary = float(input("\nEnter salary: "))
bonus = float(input("Enter bonus amount: "))

final_salary = salary + bonus

print(f"Salary after bonus: {final_salary:.2f}")


# 7. Monthly salary from annual salary

annual_salary = float(input("\nEnter annual salary: "))

monthly_salary = annual_salary / 12

print(f"Monthly Salary: {monthly_salary:.2f}")


# 8. Total cost = price × quantity + tax

price = float(input("\nEnter product price: "))
quantity = int(input("Enter quantity: "))
tax = float(input("Enter tax amount: "))

total_cost = (price * quantity) + tax

print(f"Total Cost: {total_cost:.2f}")


# 9. Distance travelled = speed × time

speed = float(input("\nEnter speed: "))
time = float(input("Enter time: "))

distance = speed * time

print(f"Distance Travelled: {distance:.2f}")


# 10. BMI using weight and height

weight = float(input("\nEnter weight in kilograms: "))
height = float(input("Enter height in meters: "))

bmi = weight / (height ** 2)

print(f"BMI: {bmi:.2f}")
```
