Ternary Operator

The ternary operator is a short way to write a simple if-else condition in one line.

Syntax


value_if_true if condition else value_if_false

```python
age = 20

result = "Adult" if age >= 18 else "Minor"

print(result)
```

```python
num = 7

result = "Even" if num % 2 == 0 else "Odd"
print(result)
```

```python
marks = 65

result = "Pass" if marks >= 40 else "Fail"
print(result)
```

```python
x=int(input())
checking="Positive" if x>=0 else "Negative"
print(checking)
```

```python
a = 25
b = 15

greater = a if a > b else b

print("Greater:", greater)
```
