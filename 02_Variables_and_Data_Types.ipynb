
Implicit Conversion

Python automatically converts some types during operations.

```python
a = 10
b = 2.5

result = a + b

print(result)
print(type(result))
```

Explicit Conversion

The programmer manually converts the type.

```python
number = "100"

integer_number = int(number)
float_number = float(number)

print(integer_number)
print(float_number)
```

```python
print(10 + 20)
print("10" + "20")#concatination
```
