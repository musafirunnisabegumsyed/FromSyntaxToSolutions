A loop is a programming structure used to repeat a block of code multiple times until a specified condition is met or the sequence is completed

While Loop

Initialization → Condition → Execution → Increment/Decrement → Repeat

```python
#Print numbers from 1 to 5
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
#Print numbers from 5 to 1
i = 5

while i >= 1:
    print(i)
    i -= 1
```

```python
#Print even numbers from 1 to 10
i = 2

while i <= 10:
    print(i)
    i += 2
```

```python
#Print odd numbers from 1 to 10
i = 1

while i <= 10:
    print(i)
    i += 2
```

```python
#. Print multiples of 5 from 5 to 50
i = 5

while i <= 50:
    print(i)
    i += 5
```

```python
#Find the sum of numbers from 1 to 10
i = 1
total = 0

while i <= 10:
    total += i
    i += 1

print("Sum:", total)
```

```python
#Print the multiplication table of a number
n = int(input("Enter a number: "))

i = 1

while i <= 10:
    print(n, "x", i, "=", n * i)
    i += 1
```

For loop

for variable in sequence:

    # code to execute

```python
#for with range()
for i in range(1, 6):
    print(i)
```

```python
#for with range(start, stop)
for i in range(1, 6):
    print(i)
```

```python
#for with range(start, stop, step)
for i in range(2, 11, 2):
    print(i)
```

```python
#for loop with a String
name = "Begum"

for char in name:
    print(char)
```

```python
#for loop with a List
numbers = [10, 20, 30, 40]

for number in numbers:
    print(number)
```

```python
#for loop with a Tuple
numbers = (10, 20, 30)

for number in numbers:
    print(number)
```

```python
#for loop with a Set
fruits = {"apple", "banana", "mango"}

for fruit in fruits:
    print(fruit)
```

```python
#for loop with a Dictionary
student = {
    "name": "Ali",
    "age": 20,
    "city": "Hyderabad"
}

for key in student:
    print(key)
```

```python
#for loop through Dictionary Key-Value Pairs
for key, value in student.items():
    print(key, value)
```

```python
#for loop through Dictionary Values
for value in student.values():
    print(value)
```

```python
for i in range(1, 11):
    if i % 2 == 0:
        print(i)
```
