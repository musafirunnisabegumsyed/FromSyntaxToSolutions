Output Formatting

f before the string → tells Python it's an f-string

{name} → inserts the value stored in name

{age} → inserts the value stored in age

```python
name = "Musafirunnisa"
age = 21

print("Name:", name)
print("Age:", age)

print(f"My name is {name}")
print(f"I am {age} years old")
```

```python
price = 1250.5678

print(f"Price: {price:.2f}")
```

```python
name = "Python"

print(f"{name:<10}")
print(f"{name:>10}")
print(f"{name:^10}")
```

Swapping Variables

Using a temporary variable

```python
a = 10
b = 20

temp = a
a = b
b = temp

print("a =", a)
print("b =", b)
```

```python
a = 10
b = 20

a, b = b, a

print("a =", a)
print("b =", b)
```
