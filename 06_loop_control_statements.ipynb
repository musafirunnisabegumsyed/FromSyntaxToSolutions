break

break immediately stops the loop.

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

continue

continue skips the current iteration and moves to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

pass

pass does nothing. It is used as a placeholder when Python requires a statement but you don't want to execute anything yet.

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

| Statement  | What it does                |
| ---------- | --------------------------- |
| `break`    | Stops the entire loop       |
| `continue` | Skips the current iteration |
| `pass`     | Does nothing                |
