Parameters

A parameter is a variable written inside the function definition.

```python
def greet(name):
    print("Hello", name)

greet("Musafir")
```

name → parameter

"Musafir" → argument

```python
def add(a, b):
    print(a + b)

add(10, 20)
```

Arguments

An argument is the actual value passed to a function when calling it.


```python
def greet(name):
    print("Hello", name)

greet("Alice")
```

Positional Arguments

Arguments are assigned to parameters based on their position.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student("Alice", 20)
```

Keyword Arguments

You explicitly specify the parameter name.

```python
def student(name, age):
    print("Name:", name)
    print("Age:", age)

student(age=20, name="Alice")
```

Default Arguments

A parameter can have a default value.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Alice")
```

```python
def calculate_bill(amount, tax=18):
    print("Amount:", amount)
    print("Tax:", tax)

calculate_bill(1000)
calculate_bill(1000, 10)
```

return

return sends a value back from the function.

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

*args

*args allows a function to accept any number of positional arguments.

args is stored as a tuple.

```python
def add_numbers(*args):
    print(args)

add_numbers(10, 20, 30, 40)
```

```python
def add_numbers(*args):
    total = 0

    for num in args:
        total += num

    return total

print(add_numbers(10, 20))
print(add_numbers(10, 20, 30, 40))
```

**kwargs

**kwargs allows a function to accept any number of keyword arguments.

**kwargs is stored as a dictionary.

```python
def student_info(**kwargs):
    print(kwargs)

student_info(name="Alice", age=20, city="Hyderabad")
```

```python
def student_info(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

student_info(
    name="Alice",
    age=20,
    course="CSE"
)
```
